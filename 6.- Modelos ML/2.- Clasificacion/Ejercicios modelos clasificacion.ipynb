```python
# detect categorical columns

def get_categorical_columns(df):
    categorical_columns = [
        col for col in df.columns
        if df[col].dtype == 'object' and 2 <= df[col].nunique() <= 5
    ]
    return categorical_columns

get_categorical_columns(churn)
```

```python
#store the categorical columns
categorical_columns = get_categorical_columns(churn)
```

```python


def encode_categorical_columns(df, categorical_columns):
    for col in categorical_columns:
        df[col + '_enc'], _ = pd.factorize(df[col])
    return df

churn=encode_categorical_columns(churn, categorical_columns)
```

```python
# enocde the categorical columns with numbers and store them in new columns with _enc appended to the original name, store those new encoded columns in encoded_categorical_columns

def encode_categorical_columns(df, categorical_columns):
    encoded_columns = []
    for col in categorical_columns:
        encoded_col = col + '_enc'
        df[encoded_col], _ = pd.factorize(df[col])
        encoded_columns.append(encoded_col)
    return df, encoded_columns

churn, encoded_categorical_columns = encode_categorical_columns(churn, categorical_columns)
```

```python
churn.tail(15)
```

```python
print(encoded_categorical_columns)
```

```python
from sklearn.model_selection import train_test_split

# Define your target
target_column = 'Churn_enc'

# Features: all encoded columns *except* the target
X = churn[[col for col in encoded_categorical_columns if col != target_column]]
y = churn[target_column]

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.60, random_state=0)

# Check shape
print("X_test shape:", X_test.shape)
```

```python
# Initialize the model 
rf_model = RandomForestClassifier(random_state=0)

# Train the model
rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
# evaluate the model
# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Classification report
print("Classification Report:\n", classification_report(y_test, y_pred))

# Confusion matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
# Accuracy on training data
y_train_pred = rf_model.predict(X_train)
train_accuracy = accuracy_score(y_train, y_train_pred)

# Accuracy on test data
y_test_pred = rf_model.predict(X_test)
test_accuracy = accuracy_score(y_test, y_test_pred)

# Print comparison
print(f"Training Accuracy: {train_accuracy:.4f}")
print(f"Test Accuracy:     {test_accuracy:.4f}")
```
