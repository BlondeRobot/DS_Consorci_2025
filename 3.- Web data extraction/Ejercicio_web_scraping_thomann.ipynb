```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
```

```python
# Iniciar el navegador
driver = webdriver.Chrome()

# URL principal de Carrefour Supermercado
url = "https://www.thomann.de/es/cat.html"
# url = "https://www.thomann.de/es/index.html"
driver.get(url)

time.sleep(3)

links_elements = driver.find_elements(By.TAG_NAME, "a")
# Lista para guardar los enlaces únicos
links = set()

# Recorrer los elementos y obtener el texto y el href
for link in links_elements:
    try:
        url = link.get_attribute("href")
        if url:  # Solo guardar si el enlace no está vacío
            links.add(url)
    except:
        continue

# Cerrar el navegador
driver.quit()
```

```python
palabras_clave = ['guitarras','bajo','drum','teclado','amplificad','piano','organo','microf','software']
articulos = []
for clases in palabras_clave:
    for link in sorted(links):  # Ordenar alfabéticamente
        if clases in link:
            
            articulos.append(link)
articulos = set(articulos)
print(len(articulos))
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36"
}
df = pd.DataFrame([])
responses = []
soups=[]
for url in articulos:
    print(url)
    # Realizar la solicitud HTTP
    response = requests.get(url, headers=headers)
    time.sleep(3) 
    responses.append(response)
    try:
        soup = BeautifulSoup(response.text, "html.parser")
        soups.append(soup)
    except:
        print(response.status_code)
```
